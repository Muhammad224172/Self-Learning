# Multiple and Multi level Ineritance 

- Multiple Inheritance = Inherit from more than one parent class:  C(A, B)

- multilevel Inheritance = Inherit froma parent ehih inherits from another parent :  C(B) <- B(A) <= A


```python
class Animal: 

    def __init__(self , name ): 

        self.name = name 
        self.is_alive = True 

    def eat(self): 
        print(f"{self.name} is eating")

    def sleep(self): 
        print(f"{self.name} is sleeping")

class Prey(Animal): 
    

    def eat(self):
        print(f"{self.name}  is eating ")
    
    def sleep(self): 

        print(f"{self.name} is sleeping")

    def flee(self): 
        print(f"{self.name} is fleeing") 


class Preditor(Animal): 
    def hunt(self): 
        print(f"{self.name} is hunting") 



class Rabbit(Prey): 
    pass

    

class Hawk(Preditor): 
    pass 

class Fish(Preditor , Prey):  #multiple Inheritance  both prey and preditor 
    pass 
```

# Abstract Classes

- A class that cannot be instantiated on its own; Meant to be subclassed. 
- They can contain abstract methds , ehich are decleared but no implementation 

- Benefits 
1. Prevents instantiation of the lass itself 
2. Requires children to use inherited abstract methods 


```python
from abc import ABC , abstractmethod 

class Vehicle(ABC): #searves as a check and balance 
    
    @abstractmethod
    def go(self): #declear put not define  
        pass 
    
    @abstractmethod
    def stop(Self):
        pass 

class Car(Vehicle):
    def go(self):
        print("You drive the car")

    def stop(Self): 
        print("You stop the car")


class Motor(Vehicle): 
    def go(Self): 
        print("You ride the motorcyle ")

    def stop(Self): 
        print("You stop the motorcyle")

class Boat(Vehicle): 
    def go(Self): 
        print("You Started the boat")

    def stop(Self): 
        print("You Anchored the boat ")
        
```

```python
boat = Boat()
boat.go()
```

# Super()

- The fuction used in a child class to class to call methods from a parent class (superclass)
- Allows you to extend the fuctionality of inherites methods 

- Method Overriding 

- super() to extend fuctionality of a method

```python
class Shapes: 
    
    def __init__(self , color , is_filled ): 
        self.color = color
        self.is_filled = is_filled
    def Describe(self): 
        print(f"it is {self.color} color and {"filled" if self.is_filled else "not filled"}")

class Circle(Shapes) : 
    
    def __init__(self , color , is_filled , radius): 
        super().__init__(color , is_filled)
        self.radius = radius  
    
    def Describe(self):
       super().Describe() # The original method of the parent
       print(f"it is a circle with an area of {3.14 * self.radius**2} cm^2")
       

class Square(Shapes) : 
    def __init__(self , color , is_filled , width): 
        super().__init__(color , is_filled)
        self.width = width   
    def Describe(self):
       super().Describe() # The original method of the parent
       print(f"it is a Square with an area of {self.width**2} cm^2")
      

class Triangle(Shapes) : 

    def __init__(self , color , is_filled , width , height ): 
        super().__init__(color , is_filled)
        self.width = width  
        self.height = height
    def Describe(self):
       super().Describe() # The original method of the parent
       print(f"it is a Triangle with an area of {self.width * self.height * 0.5} cm^2")
    
```

```python
circle = Circle("red" , is_filled = True , radius = 5)

circle.Describe()


```

```python
Triangle = Triangle("red" , is_filled = True , width = 5 , height= 10)

Triangle.Describe()
```

## Polymorphism

- Many forms 


How to Achieve Polymorphism in python 
- An object taking many forms 

- Inheritancee and Duck typing 

```python
from abc import ABC , abstractmethod

class Shape: 
    @abstractmethod
    def area(self): 
         pass 

class circle(Shape): 
     def __init__(self , radius): 
          self.radius = radius 
     def area(self): 
          return 3.14 * self.radius * self.radius      

class square(Shape): 
     def __init__(self ,  side): 
          self.side = side  

     def area(self):
       return self.side ** 2

class triangle(Shape): 
     
     def __init__(self , side , base): 
          self.base = base 
          self.side = side 
     def area(self): 
          return 0.5 * self.side * self.base 
     
     
class pizza(circle):
     def __init__(self , topping , radius): 
          super().__init__(radius) 
          self.topping = topping
```

```python
square = square()
```

```python
shapes = [circle(8) , square(7) , triangle(2 , 7) , pizza(radius = 4 , topping= "Peperoni")]

for shape in shapes : 
    print(shape.area())
```

```python
food = pizza(radius = 4 , topping= "Peperoni")

food.area()
```

# Aggregation 

- One object has relationship with one or more other parts of Independent objects 

- Whole and parts 

```python
class Library: #container 
    def __init__(self , name):
        self.name = name  
        self.books = []

    def add_books(self , book): 
        self.books.append(book)

    def list_books(self): 
        return [f"{book.title} by {book.author}" for book in self.books]

class Book: # parts 
    def __init__(self ,title , author): 
        self.title = title 
        self.author = author 
```

# Composition 

- The composed object owns its components which cannot exsit independently 

```python
class Engine: 

    def __init__(self , horse_power ):

        self.horse_power = horse_power


class Wheel: 
    def __init__(self , size): 
        self.size = size 
     

class Car: 
    def __init__(self , make , model , horse_power , wheel_size):
        self.make = make 
        self.model = model 
        self.engine = Engine(horse_power)
        self.wheels = [Wheel(wheel_size) for wheel in range(4)]

    def details(self):
        return f"{self.make} {self.model} {self.engine.horse_power} (hp) {self.wheels[0].size} inches"
```

```python
car = Car(make = "Ford" , model = "Mustang" , horse_power= 500 , wheel_size=18)

car1 = Car(make = "Mercedes" , model = "S class" , horse_power= 900 , wheel_size=12)

print(car.details())

print(car1.details())
```

# Nested Classes 

- A class defined withint another class 

#### Benefits 

- Allows Logical groupings 
- Encapsulates private details that are not relevany outside that class 
- Keeps namespace clean and reduces possibility of naming conflicts 

```python
class Company:
    class Employee: 
        def __init__(self , name , position): 
            self.name = name 
            self.position = position

        def details(self):
            return f"{self.name}  {self.position}"

        

    def __init__(self , company_name): 
        self.company_name = company_name
        self.employees = []

    def add_employee(self , name , position):
        new_employee = self.Employee(name , position)
        self.employees.append(new_employee)

    def list_employees(self): 
        return [employee.details() for employee in self.employees]
```

```python
company = Company("Krusty Crab")
company2 = Company("Chumbucket")

company2.add_employee("Sheldon" , "Manager")
company2.add_employee("Karen" , "Assistant")


company.add_employee("Eugene" , "Manager")
company.add_employee("Spongebob" , "Cook")
company.add_employee("Squidward" , "Cashier")
```

# Static method 

- A method that belongs to a class rather than any object from that class (instance) usually used for general uutility. No access to class data.

- Instance Method : operations for instances of a class 

```python
class Employee: 

    def __init__(self , name , position): 
        self.name = name 
        self.position = position

    def get_info(self): #Instance method 
        return f"{self.name} = {self.position}"

    @staticmethod #static method 
    def is_valid_poition(position):
        valid_positions = ["Manager" , "Cshier" , "Cook" , "Janitor"] 
        return position in valid_positions
```

```python
employee = Employee("Eugene" , "Cook")
employee.is_valid_poition("Cook")
```

# Class methods 

- Allows operations related to the class itself 

- takes (cls)

- class level data such as class level variables 

```python
class Student: 

    count = 0
    total_gpa = 0

    def __init__(self , name , gpa): 
        self.name = name 
        self.gpa = gpa
        Student.count += 1
        Student.total_gpa += gpa #accumulates all gpa 

    #Instance method
    def get_info(self): 
        return f"{self.name} {self.gpa}"
    
    @classmethod
    def get_count(cls): 
        return f"Total number of student: {cls.count}"
    @classmethod
    def get_average_gpa(cls):
        if cls.count == 0:
            print("No students yet")
        else: 
            return f"{cls.total_gpa / cls.count:.2f}"
```

```python
Student1 = Student("Spongebob" , 3.2)
Student2 = Student("Patrick" , 2.5)
Student3 = Student("Sandy" , 3.9)
```

# Magic Methods 

- dunder methods 

- Authomatically called 

- we can define and customize those objects 

```python
class Book: 

    def __init__(self , title , author , num_pages): 
        self.title = title 
        self.author  = author 
        self.num_pages = num_pages

    def __str__(self):
        return f"{self.title} by {self.author}"
    
    def __eq__(self , other): 
        return self.title == other.title and self.author == other.author
    
    def __lt__(self , other): 
        return self.num_pages < other.num_pages
    
    def __gt__(self , other): 
        return self.num_pages > other.num_pages
    
    def __add__(self , other): 
        return self.num_pages + other.num_pages
    
    def __contains__(self ,keyword):
        return keyword in self.title or keyword in self.author
    
    def __getitem__(self ,key): 
        if key == "title": 
            return self.title
        elif key == "author": 
            return self.author
        elif key == "num_pages": 
            return self.num_pages
        else: 
            return f"Key {key} was not found"

    
```

# Property Decorator 

- define a method as a property

- getter , setter and deleter methods 

```python
class Rectangle: 
    def __init__(self , width , height): 
        self._width = width
        self._height = height

    @property 
    def width(self): 
        return f"{self._width:.1f} cm" 

    @property
    def height(self): 
        return f"{self._height:.1f} cm" 
    
    @width.setter 
    def width(self , new_width): 
        if new_width > 0:
            self._width = new_width
        else:
            print("Width must be greater than zero")

    @height.setter 
    def height(self , new_height): 
        if new_height > 0:
            self._height = new_height
        else:
            print("Width must be greater than zero")
            
```

```python
rectangle = Rectangle(4 ,5)


rectangle.height = 9

print(rectangle.height)
```
