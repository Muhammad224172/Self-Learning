# List 

List is a collection data type

```python
myList = ["banana " , "cherry" , "apple" ]

myList 
```

```python
myList.insert(1 , "Blueberry")


myList.remove("cherry")

print(myList)
```

```python
myList = [1 ,23,44,5]
```

```python
#sorts in_list .   
myList.sort()
print(myList)
#sorted(myList) #does not sort in place 
```

```python
myList.reverse()
myList 
```

```python
myList.clear()
myList 
```

```python
myList = ["Berry" , "Cherry"] * 5
myList[2:5:2] #start : stop : step  #negative will reverse your list 
```

```python
for i in myList: 
    if i == "apple": 
        print(f"Yes we have an {i}")
```

```python
myList = list()

myList 
```

```python
#A list can have multiple Data types 
myList  = [True , 50 , "Number"]
myList 
```

```python
list_org = ["Banana" , "cherry" , "apple"]

list_cpy = list_org #This assignment makes both. so be careful  

list_cpy.append("Lemon")

print(list_cpy , list_org)
```

```python
#List comprehension 
a = [1 ,2 , 4, 5, 5]
b = [i*i for i in a]  #expresion then a foe in Loop over the list 
b
```

# Tuple 


A collection data type similar to a List with the difference being that a Tuple cannot be changed. 



```python
mytuple = ("Max" , 28 , "Boston")
mytuple
```

```python
mytuple = ("Max",)
mytuple
```

```python
mytuple = tuple(["Max" , 28 , "Bostom"])
mytuple
```
